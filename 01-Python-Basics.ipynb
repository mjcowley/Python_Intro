## 1. Python as a Calculator

The key idea in programming is the **variable**, a name that stores a value so you can use it later. You create one with `=`.

````python
redshift = 0.5
H0 = 70          # Hubble constant in km/s/Mpc
print(redshift)
print(H0)
````

The basic mathematical operators are:

````python
print("Addition:      ", 5 + 3)
print("Subtraction:   ", 5 - 3)
print("Multiplication:", 5 * 3)
print("Division:      ", 5 / 3)
print("Power:         ", 5 ** 3)    # note: ** not ^
print("Floor division:", 5 // 3)    # divide and round down
print("Remainder:     ", 5 % 3)     # what's left over after division
````

## 2. Data Types

Every value in Python has a **type**. The most common are:

| Type | Example | Used for |
|------|---------|----------|
| `int` | `42` | Whole numbers, such as counts or IDs |
| `float` | `0.523` | Decimal numbers, such as most measurements |
| `bool` | `True` | Yes/no values |
| `str` | `"NGC 1300"` | Text |

Use `type()` to check what you've got.

````python
n_galaxies = 42
redshift = 0.523
is_detected = True
name = "NGC 1300"

print(type(n_galaxies))
print(type(redshift))
print(type(is_detected))
print(type(name))
````

You can convert between types with `int()`, `float()`, `str()` and `bool()`. This matters a lot when reading data from files, because numbers often arrive as text.

````python
z_text = "0.523"            # a string that looks like a number
z = float(z_text)           # now it's an actual number
print(z * 2)

print(int(3.9))             # int() chops off the decimal part, it doesn't round
print(round(3.9))           # use round() if you want rounding
````

### A warning about floats

Computers store decimal numbers in binary, so most can't be represented exactly. This is normal and happens in every language, but it catches everyone out once.

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

Never test whether two floats are *exactly* equal. Check that they're close instead, for example `abs(a - b) < 1e-9`. You'll see this in the check cells throughout these notebooks.

## 3. Strings

Strings hold text, such as object names, filter names and file paths. You can use single or double quotes.

### f-strings

The most useful string skill is the **f-string**: put an `f` before the quotes and anything inside `{}` is replaced by its value. You can also control how numbers are formatted.

````python
name = "M87"
distance_mpc = 16.4
mass_bh = 6.5e9

print(f"{name} is {distance_mpc} Mpc away")
print(f"Distance to 3 decimal places: {distance_mpc:.3f} Mpc")
print(f"Black hole mass: {mass_bh:.2e} solar masses")    # scientific notation
print(f"Right-aligned in 10 spaces: |{name:>10}|")      # handy for tables
````

### Useful string methods

Strings come with built-in tools, called with a dot after the string.

````python
filename = "zfourge_cdfs_catalogue.fits"

print(filename.upper())
print(filename.replace(".fits", ".csv"))
print(filename.endswith(".fits"))
print(filename.split("_"))      # split into a list of pieces
print(len(filename))            # number of characters
````

### Indexing and slicing

Each character in a string has a position, or **index**, counting from **0**. Negative indices count back from the end, so `-1` is the last character.

| Character | `N` | `G` | `C` | ` ` | `1` | `3` | `0` | `0` |
|-----------|-----|-----|-----|-----|-----|-----|-----|-----|
| Index | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |
| Negative index | -8 | -7 | -6 | -5 | -4 | -3 | -2 | -1 |

A **slice** `[start:stop]` takes everything from `start` up to, but **not including**, `stop`.

````python
name = "NGC 1300"
print(name[0])      # first character
print(name[-1])     # last character
print(name[0:3])    # characters 0, 1 and 2
print(name[4:])     # from index 4 to the end
````

### Example: sky coordinates

Positions on the sky are often given as text in sexagesimal format. Right ascension (RA) is in hours:minutes:seconds, and there are 24 hours of RA in 360°, so each hour is 15°.

The ZFOURGE CDFS field is centred near RA = `03:32:30`. Let's convert that to degrees.

````python
ra_str = "03:32:30"

parts = ra_str.split(":")       # ['03', '32', '30'], still strings
hours = float(parts[0])
minutes = float(parts[1])
seconds = float(parts[2])

ra_deg = (hours + minutes / 60 + seconds / 3600) * 15
print(f"RA = {ra_str} = {ra_deg:.4f} degrees")
````

## 4. Lists and Tuples

### Lists

A **list** holds several values in order, inside square brackets. It's the simplest way to store a set of measurements.

````python
redshifts = [0.52, 1.13, 0.87, 2.41, 0.33, 1.76]

print(redshifts)
print(len(redshifts))       # how many items
print(redshifts[0])         # first item (index 0, just like strings)
print(redshifts[-1])        # last item
print(redshifts[1:4])       # a slice: items 1, 2 and 3
````

Python has built-in functions that work on lists of numbers:

Unlike strings, lists can be changed after you create them:

````python
redshifts.append(3.05)      # add an item to the end
redshifts[0] = 0.55         # replace the first item
print(redshifts)
````

A list can hold any type, including strings or other lists. In practice, keep each list to one kind of thing.

````python
filters = ["U", "B", "V", "R", "I", "J", "H", "Ks"]
print(f"We have {len(filters)} filters, from {filters[0]} to {filters[-1]}")
print("The near-infrared filters are", filters[5:])
````

## 6. Making Decisions

### Comparisons and booleans

Comparisons give back `True` or `False`:

| Operator | Meaning |
|----------|---------|
| `==` | equal to (two equals signs, since one `=` assigns a value) |
| `!=` | not equal to |
| `<`, `>` | less than, greater than |
| `<=`, `>=` | less than or equal to, greater than or equal to |

Combine them with `and`, `or` and `not`.

````python
z = 1.4
print(z > 1)
print(z > 1 and z < 2)
print(1 < z < 2)            # Python lets you chain comparisons like this
print(not z > 1)
````

### `if`, `elif` and `else`

An `if` statement runs code only when a condition is `True`. Note the **colon** at the end of each condition, and the **indentation** (4 spaces) of the code that belongs to it. Python uses indentation to know which lines are inside the `if`.

````python
z = 1.4

if z < 1:
    print("Low redshift")
elif z < 3:
    print("Intermediate redshift")
else:
    print("High redshift")

print("This line isn't indented, so it always runs")
````

### Example: is this galaxy star-forming or quiescent?

One of the most common tools in our group is the **UVJ diagram**. Plotting a galaxy's rest-frame $U-V$ colour against its $V-J$ colour separates galaxies that are still forming stars from those that have stopped (quiescent galaxies). Dust makes star-forming galaxies redder in both colours, whereas old stars make quiescent galaxies red in $U-V$ but not in $V-J$, so the two populations land in different places.

A galaxy is classed as quiescent if all three of these are true (one common version, from [Williams et al. 2009](https://ui.adsabs.harvard.edu/abs/2009ApJ...691.1879W)):

$$U-V > 1.3, \qquad V-J < 1.6, \qquad U-V > 0.88\,(V-J) + 0.59$$

The exact boundary changes a little with redshift, but we'll use this one throughout. You'll make real UVJ diagrams in later parts.

````python
u_v = 1.8
v_j = 1.1

if u_v > 1.3 and v_j < 1.6 and u_v > 0.88 * v_j + 0.59:
    print("Quiescent")
else:
    print("Star-forming")
````

## 7. Loops

### `for` loops

A `for` loop runs the same code once for each item in a list (or any collection). As with `if`, the indented lines are inside the loop.

````python
redshifts = [0.52, 1.13, 0.87, 2.41]

for z in redshifts:
    print(f"z = {z}, light was emitted when the Universe was 1/{1 + z:.2f} of its current size")
````

`range(n)` gives the numbers `0, 1, ..., n-1`, which is useful when you just need to repeat something a set number of times.

````python
for i in range(5):
    print(i)
````

### Looping over two lists at once

Often you'll have several lists that line up, such as a name and a redshift for each galaxy. `zip()` loops over them together, and `enumerate()` gives you a counter as well.

````python
names = ["Galaxy A", "Galaxy B", "Galaxy C"]
redshifts = [0.52, 1.13, 0.87]

for name, z in zip(names, redshifts):
    print(f"{name} is at z = {z}")

print()

for i, name in enumerate(names):
    print(i, name)
````

### Building up a result

A very common pattern is to start with an empty list (or a total of zero) and add to it inside the loop.

````python
redshifts = [0.52, 1.13, 0.87, 2.41, 0.33, 1.76]

high_z = []                 # start with an empty list
for z in redshifts:
    if z > 1:
        high_z.append(z)

print(high_z)
print(f"{len(high_z)} of {len(redshifts)} galaxies are above z = 1")
````

### List comprehensions

That pattern is so common that Python has a shortcut for it, the **list comprehension**. Both of these give the same result as the loop above:

````python
high_z = [z for z in redshifts if z > 1]
print(high_z)

scale_factors = [1 / (1 + z) for z in redshifts]     # transform every item
print(scale_factors)
````

Use whichever you find clearer. A plain loop is always fine.

### `while` loops

A `while` loop keeps going as long as a condition is `True`. They're less common in data analysis, but useful when you don't know in advance how many steps you'll need. For example, how many times do you have to double a distance of 1 kpc to get past 1 Gpc?

````python
distance_kpc = 1
doublings = 0

while distance_kpc < 1e6:      # 1 Gpc = 1,000,000 kpc
    distance_kpc = distance_kpc * 2
    doublings = doublings + 1

print(f"{doublings} doublings, reaching {distance_kpc:,} kpc")
````

## 9. Functions

You've already used plenty of functions, like `print()`, `len()` and `math.sqrt()`. Writing your own is how you avoid copying and pasting the same code. If you find yourself doing the same calculation twice, make it a function.

A function is defined with `def`, takes **arguments** in brackets, and sends a result back with `return`. As usual, the body is indented.

### Example: Hubble's law

For nearby galaxies, distance is roughly proportional to redshift:

$$d \approx \frac{cz}{H_0}$$

where $c = 299{,}792$ km/s and $H_0 \approx 70$ km/s/Mpc. This only works for $z \ll 1$. At higher redshift you need a proper cosmological calculation, which Astropy does for you in Part 2.

````python
def hubble_distance(z, H0=70):
    """Approximate distance in Mpc for a low-redshift galaxy, using Hubble's law."""
    c = 299792.458          # speed of light in km/s
    return c * z / H0


print(hubble_distance(0.023))             # a galaxy in the Coma cluster
print(hubble_distance(0.023, H0=67.4))    # with a different value of H0
````

A few things to notice:

- `H0=70` is a **default value**. If you don't give `H0`, the function uses 70.
- The text in triple quotes is a **docstring** that explains what the function does. Run `help(hubble_distance)` to see it.
- `c` only exists inside the function. Variables created in a function are **local** and disappear when it finishes.

Functions can return more than one value, as a tuple you can unpack:

````python
def ra_to_degrees(ra_str):
    """Convert an RA string in hh:mm:ss to degrees. Also returns the hours as a float."""
    h, m, s = ra_str.split(":")
    hours = float(h) + float(m) / 60 + float(s) / 3600
    return hours * 15, hours


ra_deg, ra_hours = ra_to_degrees("03:32:30")
print(ra_deg, ra_hours)
````

## 10. Errors and Debugging

Everyone's code breaks, all the time. The skill isn't avoiding errors, it's fixing them quickly. Python's error messages are actually very helpful once you know how to read them.

There are two main kinds of error:

- **Syntax errors**: the code isn't valid Python, so it can't run at all. Usually a missing bracket, colon or quote.
- **Exceptions**: the code is valid, but something goes wrong while it runs.

````python
# A syntax error: missing closing bracket
print("Hello, Universe"
````

````python
# An exception: the code is valid, but it fails when it runs
redshifts = [0.52, 1.13, 0.87]
print(redshifts[3])
````

### Reading a traceback

When an exception happens, Python prints a **traceback**. Read it from the **bottom up**:

1. The **last line** tells you the type of error and a short description, such as `IndexError: list index out of range`.
2. The lines above it show **where** it happened, with an arrow pointing at the line.

Common exceptions you'll meet:

| Exception | Usual cause |
|-----------|-------------|
| `NameError` | You used a variable that doesn't exist yet. Check the spelling, or that you ran the cell that creates it |
| `TypeError` | Wrong type for an operation, such as adding a string to a number |
| `ValueError` | Right type, but an impossible value, such as `float("abc")` or `math.log10(-1)` |
| `IndexError` | Asking for a list position that doesn't exist |
| `KeyError` | Asking a dictionary for a key it doesn't have |
| `ZeroDivisionError` | Dividing by zero |
| `ModuleNotFoundError` | The library isn't installed, or the name is misspelt |
| `IndentationError` | The indentation doesn't line up |

**The worst bugs don't cause errors at all.** Code that runs but gives the wrong answer is much harder to spot, which is why you should always sanity-check your results. Is that distance sensible? Is that magnitude in the right range?

### Handling errors

Sometimes you expect something might fail, like a missing value in a catalogue. `try` and `except` let you deal with it instead of crashing.

````python
flux_values = ["12.5", "8.1", "--", "15.3"]     # "--" marks a missing value

for f in flux_values:
    try:
        print(float(f))
    except ValueError:
        print(f"Couldn't convert {f!r}, skipping")
````

Use `try` and `except` sparingly. Hiding errors you didn't expect makes bugs much harder to find. In Part 3 you'll see better ways to handle missing data.

### Debugging tips

- **Read the last line of the error first.** It usually tells you exactly what's wrong.
- **Print things.** Put `print()` statements in to see what your variables actually hold, and use `type()` to check their types.
- **Test small pieces.** Run one line at a time to find which part fails.
- **Search the error message.** Someone has almost certainly hit the same error before.
